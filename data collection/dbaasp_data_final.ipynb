Code for length distribution based class tagging

Reference source on peptide length classification:

Following paragraph is copied as it is from - "https://pmc.ncbi.nlm.nih.gov/articles/PMC9082820/"

Peptide length:
The size distribution of natural AMPs in the APD is shown in Figure 1B. The shortest peptide comprises two amino acids (aa) plus chemical modifications. The upper limit is defined by the peptides collected into the APD (currently less than 200 amino acids). Based on peptide length, they may be further grouped: ultrashort (2–9 aa), short (10–24 aa), medium (25–50 aa), and long (50–100 aa). The majority of AMPs (88%) are small with less than 50 amino acids (with a peak at 21–30 aa in Figure 1B). Approximately 10% contain 50–100 amino acids. AMPs with greater than 100 aa are antimicrobial proteins.

```python
import numpy as np

conditions = [
    (df['Length_of_Sequence'] >= 2) & (df['Length_of_Sequence'] <= 9),
    (df['Length_of_Sequence'] >= 10) & (df['Length_of_Sequence'] <= 24),
    (df['Length_of_Sequence'] >= 25) & (df['Length_of_Sequence'] <= 50),
    (df['Length_of_Sequence'] > 50) & (df['Length_of_Sequence'] <= 100),
    (df['Length_of_Sequence'] > 100)
]


choices = [
    'Ultrashort (2–9 aa)',
    'Short (10–24 aa)',
    'Medium (25–50 aa)',
    'Long (50–100 aa)',
    'Antimicrobial Protein (>100 aa)'
]


df['Length_Class'] = np.select(conditions, choices, default='Out of range (<2 aa)')
```

```python
import re

MULTI_SPLIT = re.compile(r"\s{2,}")

def split_multi(value):
    """Turn one raw multi-value DBAASP field into a list of tokens."""
    if value is None or str(value).strip() == "":
        return []
    return [p.strip() for p in MULTI_SPLIT.split(str(value).strip()) if p.strip()]

ACTIVITY_LABELS = {
    "Gram+":          "antibacterial_gram_pos",
    "Gram-":          "antibacterial_gram_neg",
    "Fungus":         "antifungal",
    "Virus":          "antiviral",
    "Cancer":         "anticancer",
    "Biofilm":        "antibiofilm",
    "Parasite":       "antiparasitic",
    "Insect":         "insecticidal",
    "Protista":       "antiprotozoal",
    "Mollicute":      "antimollicute",
    "Nematode":       "nematicidal",
    "Mammalian Cell": "mammalian_cytotoxic",
}

def add_activity_columns(df, raw_column="TARGET GROUP"):
    # step 1: tokenize the raw field
    tokens = df[raw_column].map(split_multi)

    # step 2: one boolean column per known activity
    for raw_token, clean_label in ACTIVITY_LABELS.items():
        df[clean_label] = tokens.map(lambda t, tok=raw_token: tok in t)

    # step 3 (optional extras): count + readable summary
    df["n_activities"] = tokens.map(len)
    df["activity_list"] = tokens.map(
        lambda t: "; ".join(ACTIVITY_LABELS.get(x, x) for x in t)
    )
    return df
```

## Tagging the combined database by kingdom

Each kingdom file (Animalia, Archaea, Bacteria, Fungi, Protista, Virus, Plantae) is a subset of the master `df` pulled from the same underlying DBAASP database, so every peptide keeps the same `ID` across every file it appears in — `ID` is the one column guaranteed to be unique and stable.

**Matching on `ID` instead of `NAME`**: the earlier `_ANIMALIA = df_kingdom_animalia['NAME'].tolist()` approach (cell above) matches on peptide name, but names in this dataset have inconsistent whitespace/formatting (e.g. `"Brevinin-2 related peptide B2RP   "`) and aren't guaranteed unique, so string matching on `NAME` will silently miss or mis-tag rows. `ID` doesn't have that problem, so we tag using `ID` instead.

```python
# One DataFrame per kingdom (reuses the variables already loaded above)
kingdom_frames = {
    'Animalia': df_kingdom_animalia,
    'Archaea':  df_kingdom_archaea,
    'Bacteria': df_kingdom_bacteria,
    'Fungi':    df_kingdom_fungi,
    'Protista': df_kingdom_protista,
    'Virus':    df_kingdom_virus,
    'Plantae':  df_kingdom_plantae,
}
```

```python
len(df_kingdom_animalia)
```

```python
# Build ID -> [kingdoms] map. A list (not a single value) because a peptide
# could in principle show up in more than one kingdom file (e.g. dual-annotated
# in the source database) - we don't want to silently overwrite that.
id_to_kingdoms = {}
for kingdom, kdf in kingdom_frames.items():
    for pid in kdf['ID']:
        id_to_kingdoms.setdefault(pid, []).append(kingdom)

# Data-quality check: flag any ID claimed by more than one kingdom file
overlaps = {pid: ks for pid, ks in id_to_kingdoms.items() if len(ks) > 1}
print(f"{len(overlaps)} IDs appear in more than one kingdom file")
if overlaps:
    print(dict(list(overlaps.items())[:10]))  # peek at a few
```

```python
# Tag the combined/master dataframe
df['KINGDOM'] = df['ID'].map(lambda pid: ', '.join(id_to_kingdoms.get(pid, [])))
df.loc[df['KINGDOM'] == '', 'KINGDOM'] = 'Unclassified'

print(df['KINGDOM'].value_counts())
```

`Unclassified` rows are peptides present in the master `df` slice(s) you've loaded but not present in any of the kingdom file(s) you've loaded yet — as you add the remaining kingdom pagination files (e.g. the rest of Animalia beyond `0-2645`, or any kingdom you haven't pulled in), re-run the two cells above and the `Unclassified` count will keep shrinking. It should not be treated as "kingdom = none" until every kingdom file has been loaded.

```python
import requests, concurrent.futures as cf, pandas as pd, os
from requests.adapters import HTTPAdapter

API_URL = "https://dbaasp.org/api/v1"
CHECKPOINT = "/content/kingdom_lookup.csv"
WORKERS = 60

session = requests.Session()
session.mount("https://", HTTPAdapter(pool_connections=WORKERS, pool_maxsize=WORKERS))
# ^ requests defaults pool_maxsize to 10 — without raising it, 60 threads bottleneck on 10 connections

def fetch(pid, timeout=8, retries=2):
    for _ in range(retries + 1):
        try:
            r = session.get(API_URL, params={"query": "peptide_card", "peptide_id": int(pid), "format": "json"}, timeout=timeout)
            if r.ok:
                d = r.json()
                src = d.get("source", {}) or {}
                return {"ID": pid, "KINGDOM_API": src.get("kingdom"), "SUBKINGDOM_API": src.get("subkingdom"),
                        "SOURCE_ORGANISM": src.get("source") or src.get("organism"),
                        "NCBI_ID": src.get("ncbi"), "STATUS": "ok"}
        except requests.RequestException:
            continue
    return {"ID": pid, "KINGDOM_API": None, "STATUS": "failed"}

done_ids, results = set(), []
if os.path.exists(CHECKPOINT):
    prev = pd.read_csv(CHECKPOINT)
    results, done_ids = prev.to_dict("records"), set(prev["ID"])

targets = [pid for pid in df.loc[df['KINGDOM'] == 'Unclassified', 'ID'].unique() if pid not in done_ids]

with cf.ThreadPoolExecutor(max_workers=WORKERS) as ex:
    futs = {ex.submit(fetch, pid): pid for pid in targets}
    for i, fut in enumerate(cf.as_completed(futs), 1):
        results.append(fut.result())
        if i % 1000 == 0:
            pd.DataFrame(results).to_csv(CHECKPOINT, index=False)
            print(f"{i}/{len(targets)}")

pd.DataFrame(results).to_csv(CHECKPOINT, index=False)
```

```python
import os, re, pandas as pd

OUTPUT_ROOT = "/content/subsets"
manifest = []

def slugify(value):
    value = re.sub(r'[\\/]+', '-', str(value).strip())   # "DNA / RNA" -> "DNA - RNA"
    value = re.sub(r'\s+', '_', value)
    return re.sub(r'[^\w\-]', '', value)

def write_subset(subset_df, dimension, label):
    folder = os.path.join(OUTPUT_ROOT, dimension)
    os.makedirs(folder, exist_ok=True)
    path = os.path.join(folder, f"{slugify(label)}.csv")
    subset_df.to_csv(path, index=False)
    manifest.append({"dimension": dimension, "value": label, "n_rows": len(subset_df), "file": path})
```

```python
for col, dim in [('DOMAIN', 'by_domain'), ('ORIGIN', 'by_origin'), ('Length_Class', 'by_length_class')]:
    for value, group in df.groupby(col):
        write_subset(group, dim, value)
```
